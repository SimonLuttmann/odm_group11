# Top Trumps Pipeline Steuerung (Auto-Run)

Dieses Notebook führt alle Experimente in der korrekten Reihenfolge aus.

**2-Spieler Pipeline (Standard):**
1. `01_setup_and_exploration.ipynb`: Setup & Exploration
2. `02_optimization.ipynb`: Optimierung & Validierung (Haupt-Teil)
3. `03_analysis.ipynb`: Analyse der Ergebnisse & Plotting
4. `04_algorithm_comparison.ipynb`: Vergleich verschiedener Algorithmen
5. `05_parameter_study.ipynb`: Parameter-Studie K

**Multiplayer Pipeline (Optional):**
6. `06_multiplayer_setup.ipynb`: Multiplayer Setup & Exploration
7. `07_multiplayer_optimization.ipynb`: Multiplayer Optimierung
8. `08_multiplayer_configurations.ipynb`: Vergleich: 2v1, 1v2, 2v2, etc.
9. `09_multiplayer_parameter_study.ipynb`: Parameter-Studie K für Multiplayer
10. `10_optimal_configuration.ipynb`: Optimale Spieleranzahl & Verteilung (K=22, L=4)

**Steuerung:** Setze `RUN_MULTIPLAYER = True/False` in Zelle 3

```python
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
from pathlib import Path
import time

def run_notebook(notebook_filename):
    path = Path(notebook_filename)
    print(f"--> Starte {path.name}...")
    start_time = time.time()
    
    with open(path) as f:
        nb = nbformat.read(f, as_version=4)
        
    # Führe Notebook aus
    # kernel_name="python3" stellt sicher, dass der Standard-Python-Kernel genutzt wird
    ep = ExecutePreprocessor(timeout=None, kernel_name="python3")
    
    try:
        # Führe im aktuellen Verzeichnis aus
        ep.preprocess(nb, {"metadata": {"path": "."}})
        
        # Speichere Output (inplace), damit man die Ergebnisse auch in den Dateien sieht
        with open(path, "w", encoding="utf-8") as f:
            nbformat.write(nb, f)
            
        print(f"    Fertig in {time.time() - start_time:.1f} Sekunden.")
        return True
        
    except Exception as e:
        print(f"    FEHLER bei {path.name}: {e}")
        # Bei Fehler trotzdem speichern um den Stacktrace zu sehen
        with open(path, "w", encoding="utf-8") as f:
            nbformat.write(nb, f)
        return False
```

```python
# === 2-SPIELER PIPELINE (Standard) ===
notebooks_base = [
    "01_setup_and_exploration.ipynb",
    "02_optimization.ipynb",
    "03_analysis.ipynb",
    "04_algorithm_comparison.ipynb",
    "05_parameter_study.ipynb"
]

# === MULTIPLAYER PIPELINE (Optional) ===
# Setze RUN_MULTIPLAYER = True um auch die Multiplayer-Notebooks auszuführen
RUN_MULTIPLAYER = True

notebooks_multiplayer = [
    "06_multiplayer_setup.ipynb",
    "07_multiplayer_optimization.ipynb",
    "08_multiplayer_configurations.ipynb",
    "09_multiplayer_parameter_study.ipynb",
    "10_optimal_configuration.ipynb"
]

# Kombiniere Listen basierend auf Einstellung
notebooks = notebooks_base.copy()
if RUN_MULTIPLAYER:
    notebooks.extend(notebooks_multiplayer)
    print("✓ Multiplayer-Notebooks werden auch ausgeführt\n")

print("=== Starte Pipeline ===\n")
total_start = time.time()

for nb in notebooks:
    success = run_notebook(nb)
    if not success:
        print("\nPipeline abgebrochen wegen Fehler. Prüfe das betroffene Notebook.")
        break

print(f"\n=== Pipeline beendet in {time.time() - total_start:.1f} Sekunden ===")
```
