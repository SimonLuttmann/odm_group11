# Top Trumps Pipeline Steuerung (Auto-Run)

Dieses Notebook führt alle Experimente in der korrekten Reihenfolge aus.

**Ablauf:**
1. `01_setup_and_exploration.ipynb`: Setup & Exploration
2. `02_optimization.ipynb`: Optimierung & Validierung (Haupt-Teil)
3. `03_analysis.ipynb`: Analyse der Ergebnisse & Plotting
4. `04_algorithm_comparison.ipynb`: Vergleich verschiedener Algorithmen (Optional)

```python
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
from pathlib import Path
import time

def run_notebook(notebook_filename):
    path = Path(notebook_filename)
    print(f"--> Starte {path.name}...")
    start_time = time.time()
    
    with open(path) as f:
        nb = nbformat.read(f, as_version=4)
        
    # Führe Notebook aus
    # kernel_name="python3" stellt sicher, dass der Standard-Python-Kernel genutzt wird
    ep = ExecutePreprocessor(timeout=None, kernel_name="python3")
    
    try:
        # Führe im aktuellen Verzeichnis aus
        ep.preprocess(nb, {"metadata": {"path": "."}})
        
        # Speichere Output (inplace), damit man die Ergebnisse auch in den Dateien sieht
        with open(path, "w", encoding="utf-8") as f:
            nbformat.write(nb, f)
            
        print(f"    Fertig in {time.time() - start_time:.1f} Sekunden.")
        return True
        
    except Exception as e:
        print(f"    FEHLER bei {path.name}: {e}")
        # Bei Fehler trotzdem speichern um den Stacktrace zu sehen
        with open(path, "w", encoding="utf-8") as f:
            nbformat.write(nb, f)
        return False
```

```python
notebooks = [
    "01_setup_and_exploration.ipynb",
    "02_optimization.ipynb",
    "03_analysis.ipynb",
    "04_algorithm_comparison.ipynb",
    "05_parameter_study.ipynb"
]

print("=== Starte Pipeline ===\n")
total_start = time.time()

for nb in notebooks:
    success = run_notebook(nb)
    if not success:
        print("\nPipeline abgebrochen wegen Fehler. Prüfe das betroffene Notebook.")
        break

print(f"\n=== Pipeline beendet in {time.time() - total_start:.1f} Sekunden ===")
```
